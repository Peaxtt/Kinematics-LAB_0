## RTB Practice - Forward Kinematics Warm-up

```python
import numpy as np
from roboticstoolbox import DHRobot, RevoluteMDH
from spatialmath import SE3
import swift

print("Check whether swift is installed and working")

env = swift.Swift()
env.launch(realtime=True)
```

### 1 - Build a toy 2-link planar arm
Link 1: length 0.2 m. Link 2: length 0.15 m. Both revolute, planar.

```python
toy = DHRobot([
    RevoluteMDH(a=0.0,  alpha=0.0, d=0.0),
    RevoluteMDH(a=0.15,  alpha=0.0, d=0.0),
], name='toy_2link')
toy.tool = SE3.Tx(0.20)
print(toy)
```

### 2 - Forward kinematics at a known pose
At q = [0, 0] the arm is straight along x

```python
q_zero = np.array([0.0, 0.0])
T = toy.fkine(q_zero)
print('T =\n', T)
print('tip :', np.round(T.t, 4))
```

### 3 - Try a bent pose and compare to hand calculation
Set both joints to 90 degrees, work out the expected tip position on paper

```python
q_bent = np.array([np.deg2rad(90), np.deg2rad(90)])
T = toy.fkine(q_bent)
tip = T.t
print('T =\n', T)
print('tip :', np.round(tip, 4))

# edit here: your hand-calculated (x, y, z) for q_bent
hand_calc = np.array([0.0, 0.0, 0.0])
print('diff:', round(float(np.linalg.norm(tip - hand_calc)), 4), 'm')
```

### 4 - Sweep joint 1 and watch the tip move
Loop `q1` from 0 to 90 degrees with `q2` fixed, print the tip each step.

```python
for deg in range(0, 91, 30):
    q = np.array([np.deg2rad(deg), np.deg2rad(30)])
    tip = toy.fkine(q).t
    print(f'q1={deg:3d} deg -> tip {np.round(tip, 4)}')
```
